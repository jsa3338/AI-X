```python
df = pd.read_csv('data/배추요약데이터_직팜정리.csv', encoding='cp949')
df.info()
```

```python
df['산지코드'] = df['산지코드'].astype('str')
```

```python
df.fillna({'상품 중분류 이름' : '배추'}, inplace=True)
```

```python
df_code = pd.read_csv('data/표준코드/표준코드_산지코드.csv', encoding='cp949')
```

```python
df_code.head()
```

```python
df_code.info()
```

```python
# 3. 컬럼명 통일: 기준 파일의 컬럼명을 df 기준에 맞게 변경
df_code = df_code.rename(columns={
    'plor_cd': '산지코드',
    'plor_nm': '표준_산지이름'
})


# 4. 산지코드 기준으로 merge (산지이름 채우기용)
df = df.merge(df_code, on='산지코드', how='left')

# 5. 산지이름이 비어있는 경우에만 채움
df['산지이름'] = df['산지이름'].fillna(df['표준_산지이름'])

# 6. 임시 컬럼 제거
df = df.drop(columns=['표준_산지이름'])
```

```python
df.info()
```

```python
df[df['산지이름'].isna()]
```

```python
df.dropna(axis=0, how='any', subset=None, inplace=True)
```

```python
df
```

```python
# 1. 날짜를 datetime으로 변환
df['날짜(YYYY-MM-DD)'] = pd.to_datetime(df['날짜(YYYY-MM-DD)'])

# 2. 연도-주차 컬럼 만들기
df['연도'] = df['날짜(YYYY-MM-DD)'].dt.isocalendar().year
df['주차'] = df['날짜(YYYY-MM-DD)'].dt.isocalendar().week
```

```python
df[df['주차']==53]
```

```python
# 3. 주차별 그룹화 및 합계 구하기
df_weekly = df.groupby(['연도', '주차','직팜산지코드']).agg({
    '총가격(원)': 'sum',
    '단위총물량(kg)': 'sum',
    # 필요시 다른 컬럼도 추가 가능
}).reset_index()

# 4. 연도-주차를 하나의 문자열로 붙이고 싶다면 (선택)
df_weekly['연도_주차'] = df_weekly['연도'].astype(str) + '-W' + df_weekly['주차'].astype(str).str.zfill(2)
```
