# 주차 별 합산 코드

```python
import pandas as pd
import numpy as np
```

```python
# 병합된 파일 준비
df = pd.read_csv('data/배추요약데이터_직팜정리.csv', encoding='cp949')
df.info()
```

```python
# 1. 날짜를 datetime으로 변환
df['날짜(YYYY-MM-DD)'] = pd.to_datetime(df['날짜(YYYY-MM-DD)'])

# 2. 연도-주차 컬럼 만들기
df['연도'] = df['날짜(YYYY-MM-DD)'].dt.isocalendar().year
df['주차'] = df['날짜(YYYY-MM-DD)'].dt.isocalendar().week

# 3. 주차별 그룹화 및 합계 구하기
df_weekly = df.groupby(['연도', '주차','직팜산지코드']).agg({
    '총가격(원)': 'sum',
    '단위총물량(kg)': 'sum',
    # 필요시 다른 컬럼도 추가 가능
}).reset_index()

# 4. 연도-주차를 하나의 문자열로 붙이고 싶다면 (선택)
df_weekly['연도_주차'] = df_weekly['연도'].astype(str) + '-W' + df_weekly['주차'].astype(str).str.zfill(2)
```
