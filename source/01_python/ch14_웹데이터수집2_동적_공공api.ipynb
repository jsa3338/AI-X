````python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
````

````python
driver = webdriver.Chrome()
driver.get("http://www.python.org")
elem = driver.find_element(By.NAME,'q')
elem.clear()
elem.send_keys('pycon')
elem.send_keys(Keys.RETURN)  # ENTER를 쳐도 되지만 SUBMIT이 일어날때는 주로 RETURN을 쓴다
````

````python
result_list = driver.find_elements(By.CSS_SELECTOR, 'h3 > a')
for result in result_list:
    print(f"{result.text} - {result.get_attribute('href')}")
````

````python
from bs4 import BeautifulSoup
soup = BeautifulSoup(driver.page_source, "html.parser")
result_list = soup.select('h3 > a')
for result in result_list:
    print(f"{result.text} - {domain+result.attrs['href']}")
````

````python
# soup 객체를 사용했더니 url이 full로 나오지 않음.
# 셀레니움을 통해 접근한 현재 url을 가져와서 붙이자
from urllib.parse import urlparse
current_url = driver.current_url
print('현재 주소:', current_url)
parse_url = urlparse(current_url)
print('파싱한 주소:', parse_url)
domain = f"{parse_url.scheme}://{parse_url.netloc}"
print('domain:', domain)
````

#### 페이징 처리
- 원하는 단어를 원하는 페이지 수 많큼

````python
# 다음 접속 -> 검색창 비트코인 검색 -> 뉴스 탭 -> 기사 크롤링
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
driver = webdriver.Chrome()
driver.get('https://www.daum.net/')

driver.find_element(By.NAME, 'q').click()
query = input('검색 단어 : ')
pages = int(input('페이지(1~n) : '))
driver.find_element(By.CSS_SELECTOR, 'input[type="text"]').send_keys(query)
driver.find_element(By.CLASS_NAME, 'btn_ksearch').click()
# 화면 전환시 sleep을 해주는게 좋다. 다음페이지가 뜨기도 전에 실행하면 에러가 발생함!
time.sleep(2)

# 뉴스탭 클릭
driver.find_element(By.LINK_TEXT, '뉴스').click()
time.sleep(2)

news_list = []

for page in range(1, pages+1):    
    elems = driver.find_elements(By.CSS_SELECTOR, 'div.item-title> strong.tit-g.clamp-g>a')
    for el in elems:
        title = el.text
        link = el.get_attribute('href')
        news_list.append([title, link])
    page_div = driver.find_element(By.CSS_SELECTOR, 'div.inner_paging')
    next_page = page_div.find_element(By.LINK_TEXT, str(page+1))
    next_page.click()
    time.sleep(2)
    print(f'=== {page}페이지 데이터 수집완료 ===')
driver.close()
import pandas as pd
df = pd.DataFrame(news_list, columns=['title', 'link'])
df.to_csv('data/ch14_daum.csv', index=False)
````

### 2) 맞춤법 검사기
- 네이버 맞춤법 검사기 이용
- 네이버 접속 -> 검색창 맞춤법검사기 검색 - > 검사기에 맞춤법 검사 하기 -> 검사된 결과 가져오기 
- -> 번역사이트에서 번역하기 - > 번역된 결과를 가져오기

````python
# ch14_맞춤법전.txt 를 300자로 쪼개기
with open('data/ch14_맞춤법전.txt', 'r', encoding='utf-8') as f:
    text = f.read()
print('총글자수 :', len(text))
ready_list=[]
while(len(text) >300):
    temp = text[:300]
    new_line_char_index = temp.rfind('\n')
    print(new_line_char_index)
    ready_list.append(text[:new_line_char_index])
    text = text[new_line_char_index:]
ready_list.append(text)
````

````python
# 네이버 맞춤법 검사기 -> 검사 및 크롤링 -> 파일 작성
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver = webdriver.Chrome()
time.sleep(1)
driver.get('https://www.naver.com/')
input_el = driver.find_element(By.NAME, 'query')
input_el.send_keys('맞춤법 검사기')
input_el.send_keys(Keys.ENTER)
time.sleep(1)
results = ''
for ready in ready_list:
    textarea = driver.find_element(By.CLASS_NAME, 'txt_gray')
    textarea.send_keys(ready)
    button_chk = driver.find_element(By.CLASS_NAME, 'btn_check')
    button_chk.click()
    time.sleep(1)
    
    soup =  BeautifulSoup(driver.page_source, 'html.parser')
    result = soup.select_one('p._result_text.stand_txt').text
    results += result + '\n'
    button_del = driver.find_element(By.CLASS_NAME, 'delete_btn')
    button_del.click()
    time.sleep(0.5)
driver.close()
with open('data/ch14_맞춤법후.txt', 'w', encoding='utf-8') as f:
    f.write(results)
````

### 3) 자동화 번역하기 (quiz) 

````python
# ch14_맞춤법후.txt 를 1000자로 쪼개기
with open('data/ch14_맞춤법후.txt', 'r', encoding='utf-8') as f:
    text = f.read()
ready_lists= []
while(len(text)>1000):
    temp = text[:1000]
    flag = temp.rfind('\n')
    temp = temp[:flag]
    ready_lists.append(temp)
    text = text[flag:]
ready_lists.append(text)

# 카카온 번역에서 번역 후 ch14_자동화영어번역본.txt 로 저장
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
driver = webdriver.Chrome()
driver.get('https://translate.kakao.com')
time.sleep(0.5)
textarea = driver.find_element(By.ID, 'textareaWrite')
result = ''
for ready in ready_lists:
    time.sleep(0.5)
    textarea.send_keys(Keys.CONTROL, 'a')
    textarea.send_keys(ready)
    btn_trans = driver.find_element(By.CLASS_NAME, 'btn_translate')
    btn_trans.click()
    time.sleep(0.5)
#     soup = BeautifulSoup(driver.page_source, 'html.parser')
#     result_area = soup.select_one('div.result_area').text
    result_area = driver.find_element(By.CSS_SELECTOR, 'div.result_area').text    
    result += result_area +'\n\n'
driver.close()
#번역한 결과 파일 출력
with open('data/ch14_자동화영어번역본.txt', 'w', encoding='utf-8') as f:
    f.write(result)
print('번역 완료')
````
