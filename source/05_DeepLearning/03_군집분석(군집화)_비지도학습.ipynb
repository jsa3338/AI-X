# 2. 군집화 로직

```python
# 초기 랜점 지점 2개
random_points = [[random.randint(40,100), random.randint(140,190)],
                [random.randint(40,100), random.randint(140,190)]]
random_points
```

```python
plt.plot([d[0] for d in data],
        [d[1] for d in data], '^', color='k')
plt.plot(random_points[0][0], random_points[0][1], 'x', color='r', markersize =20) # 기준점 1
plt.plot(random_points[1][0], random_points[1][1], 'x', color='b', markersize =20) # 기준점 2
```

```python
# 두 점의 거리를 리턴 (0,0) ~(4,3)
def dist(a, b):
    return np.sqrt( (a[0]-b[0])**2 + (a[1]-b[1])**2 )
```

```python
a = [0,0]; b = [4,3]
dist(a, b)
```

```python
# 두 영역(random_points[0], random_points[1])을 기준으로 나눔
g1 = []  # random_points[0]에 가까운 그룹
g2 = []  # random_points[1]에 가까운 그룹
for d in data:
    if dist(random_points[0], d) < dist(random_points[1], d):
        g1.append(d)
    else:
        g2.append(d)
len(g1), len(g2)
```

```python
# 새로운 그룹1과 그룹2, 초기 랜덤 지점을 그래프
plt.plot([g[0] for g in g1],
        [g[1] for g in g1], 'o', color='r')
plt.plot([g[0] for g in g2],
        [g[1] for g in g2], '^', color='b')

plt.plot(random_points[0][0], 
         random_points[0][1], 
         'x', color='r', markersize =20) # 기준점 1
plt.plot(random_points[1][0], 
         random_points[1][1], 
         'x', color='b', markersize =20) # 기준점 2
```

```python
sumX = 0; sumY=0
for g in g1:
    sumX += g[0]
    sumY += g[1]
# 새로운 기준점
random_points[0] = [sumX/len(g1), sumY/len(g1)]

sumX = 0; sumY=0
for g in g2:
    sumX += g[0]
    sumY += g[1]
random_points[1] = [sumX/len(g2), sumY/len(g2)]

random_points
```

```python
# 새로운 그룹1과 그룹2, 초기 랜덤 지점을 그래프
plt.plot([g[0] for g in g1],
        [g[1] for g in g1], 'o', color='r')
plt.plot([g[0] for g in g2],
        [g[1] for g in g2], '^', color='b')

plt.plot(random_points[0][0], 
         random_points[0][1], 
         'x', color='r', markersize =20) # 기준점 1
plt.plot(random_points[1][0], 
         random_points[1][1], 
         'x', color='b', markersize =20) # 기준점 2
```

```python
# 새로운 기준점을 중심으로 그룹 재편
# 두 영역(random_points[0], random_points[1])을 기준으로 나눔
g1 = []  # random_points[0]에 가까운 그룹
g2 = []  # random_points[1]에 가까운 그룹
for d in data:
    if dist(random_points[0], d) < dist(random_points[1], d):
        g1.append(d)
    else:
        g2.append(d)
len(g1), len(g2)
```

- 랜덤포인트 2지점
- 랜덤포인트 2지점 기준으로 group1과 group2

# 4. API(sklearn) 사용한 군집화
- sklearn 머신러닝
    * 예측모델 - 분류, 회귀, 군집분석
        - fit(학습),predict(예측), 
    * 변환모델
        ; 전처리 ex. 스케일조정
        - fit(학습), transform(변환함수)

```python
# 군집 분석 라이브러리가 joblib랑 충돌할 수 있음 - 경고메세지 출력
import os
os.environ["OMP_NUM_TREADS"] = "1"
```

```python
# sklearn은 기본적으로 numpy 배열로 받는다
data = np.array(data)
from sklearn.cluster import KMeans
model = KMeans(n_clusters=2,  # 2개 그룹으로 나눠
               init='random',  # 초기 중심점 랜덤하게
               n_init=10,  # 10번 실행
              )
model.fit(data)
```

```python
# 중심점
model.cluster_centers_
```

```python
# 나눠짐 그룹들의 인덱스
model.labels_
```

```python
group1 = data[model.labels_==0]
group2 = data[model.labels_==1]
c1, c2 = model.cluster_centers_
```

```python
plt.plot([g[0] for g in group1],
        [g[1] for g in group1],
        'o', c='r')
plt.plot([g[0] for g in group2],
        [g[1] for g in group2],
        'o', c='b')
plt.plot(c1[0], c1[1], 'x', c='r', markersize=15)
plt.plot(c2[0], c2[1], 'x', c='b', markersize=15)
```
