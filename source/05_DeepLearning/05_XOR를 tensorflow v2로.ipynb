```python
import numpy as np
from tensorflow.keras.models import Sequential  # Sequential 클래스의 객체 model을 만든다
from tensorflow.keras.layers import Input, Dense  # layer 층을 쌇을때 필요
```

```python
# 1. 데이터 확보 및 세팅
# 2. 데이터 전처리

X_data = np.array([[0,0],
                  [0,1],
                  [1,0],
                  [1,1]])
y_data = np.array([[0], [1], [1], [0]])
```

```python
# 3. 모델 구성
# layer층 x_data 2개 -> 10 -> 20 -> 10 -> 1
model = Sequential()
# model.add(input(shape=(2,)))  # 입력층
model.add(Dense(input_dim=2,  # 입력개수
                units=10,    # 출력개수
                activation='relu'))
model.add(Dense(units=20,    # 입력개수를 따로 입력하지 않으면 바로 윗줄의 결과물을 입력으로 받아들인다. 생략 가능!
                activation='relu'))
model.add(Dense(units=10,    
                activation='relu'))
model.add(Dense(units=1,    
                activation='sigmoid'))  
# 이진분류에서의 출력층 활성화 함수는 sigmoid
# 다중분류라면 softmax를 사용(출력합을 1로)
model.summary()
```

```python
# 4. 모델 학습과정 설정
model.compile(loss="binary_crossentropy", # 분류분석에서의 손실함수(원핫인코딩有)
             optimizer='adam',
             metrics=['binary_accuracy'])
```

```python
# 5. 학습
hist = model.fit(X_data, y_data,
                epochs=1000,
                verbose=2 # 0: 출력없음, 1:기본값, 2:에포크별 상세출력
                )
```

```python
# 5. 학습과정 
hist.history.keys()
```

```python
import matplotlib.pyplot as plt
fig, loss_ax = plt.subplots(figsize = (10,5))
loss_ax.plot(hist.history['loss'], 'r', label = 'loss')
loss_ax.set_xlabel('epochs')
loss_ax.set_ylabel('loss')
loss_ax.legend(loc='center right')
acc_ax = loss_ax.twinx()
acc_ax.plot(hist.history['binary_accuracy'], 'b', label = 'binary_accuracy')
acc_ax.set_ylabel('binary_accuracy')
```

```python
# 6. 학습평가하기
score = model.evaluate(X_data, y_data)
print('loss :', score[0])
print('binary accuracy :', score[1])
```

```python
# 7. 모델 사용
model.predict(np.array([[0,1],
                       [1,1]])).round()
```
