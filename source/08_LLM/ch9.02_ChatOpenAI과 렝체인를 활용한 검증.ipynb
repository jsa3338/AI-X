# 2. LLM 답변 생성
## 2.1 openAI SDK 를 선택
- Chat_Compeletion
- 세밀한 제어가 필요할때, 성능최적화가 중요할 때 -> 랭체인 사용 안함

```python
from openai import OpenAI
client = OpenAI()
response = client.chat.completions.create(
    model = "gpt-4.1-nano",
    messages = [
        {"role" : "system", "content": "너는 친절하게 답변해주는 비서야"},
        {"role" : "user", "content" : "2020년 월드시리즈 누가 우승했어?"}
    ]
)
```

```python
response.choices[0].message.content
```

## 2.2 LangChain을 선택
- 내부적으로 자동 메세지 형태를 변환해 줌

```python
from langchain_openai import ChatOpenAI
# .env 파일에 OPEN_API_KEY 이 등록되어있으면 아래와 같이 가능
llm = ChatOpenAI(
    model = "gpt-4.1-nano"
)
ai_message = llm.invoke("2020년 월드시리즈 누가 우승했어?")
ai_message
```

```python
ai_message.content
```
