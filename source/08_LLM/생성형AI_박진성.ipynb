```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# food_chain 구현
llm = ChatOllama(model="llama3.2:1b")
food_prompt_template = PromptTemplate(
    template = "Answer The best famous food name of the {country}. Return the food name only.",
    input_variables = ["country"]
)
output_parser = StrOutputParser()
food_chain = food_prompt_template | llm | output_parser
food_chain.invoke({"country":"Korea"})
```

```python
# recipe_chain 구현
# 페르소나 & 답변 방식 정하기
from langchain_core.prompts import ChatPromptTemplate
chatPromptTemplate = ChatPromptTemplate.from_messages([
    ('system', 'You are a food recipe master!'),  
    ('human', '''Write a comprehensive recipe for {food} including ingredients, preparation steps, and useful cooking tips.''')])
# prompt = chatPromptTemplate.invoke({'food':'Bibimbap'})
# output_parser.invoke(llm.invoke(prompt))

# recipe_chain 구현
recipe_chain = chatPromptTemplate | llm | output_parser
recipe_chain.invoke({'food':'Bibimbap'})
```

```python
# food_chain & recipe_chain 연결
final_chain = food_chain | recipe_chain 
result = final_chain.invoke({"country":"Japan"})
print(result)
```
